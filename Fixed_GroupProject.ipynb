# Assigning Groupings

```r
cats.success.outcomes = cat.outcomes %>%
  mutate(Successful.Outcome = ifelse(Outcome.Type %in% c("Return to Owner", "Adoption", "RTO Adopt"), 1, 0))
```

```r
head(cats.success.outcomes)
```

```r
library(stringr)
library(dplyr)

cats.success.outcomes.age = cats.success.outcomes %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " year", "years")) %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " month", "months")) %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " week", "weeks")) %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " day", "days"))

cats.success.outcomes.age = cats.success.outcomes %>%
  mutate(AgeInDays = case_when(
    str_detect(Age.upon.Outcome, "years") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")) * 365,
    str_detect(Age.upon.Outcome, "months") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")) * 30,
    str_detect(Age.upon.Outcome, "weeks") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")) * 7,
    str_detect(Age.upon.Outcome, "days") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")),
    TRUE ~ NA_real_
  ))
head(cats.success.outcomes.age)
```

```r
dogs.success.outcomes = dog.outcomes %>%
  mutate(Successful.Outcome = ifelse(Outcome.Type %in% c("Return to Owner", "Adoption", "RTO Adopt"), 1, 0))
```

```r
head(dogs.success.outcomes)
```

```r
dogs.success.outcomes.age = dogs.success.outcomes %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " year", "years")) %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " month", "months")) %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " week", "weeks")) %>%
  mutate(Age.upon.Outcome = str_replace_all(Age.upon.Outcome, " day", "days"))

dogs.success.outcomes.age = dogs.success.outcomes %>%
  mutate(AgeInDays = case_when(
    str_detect(Age.upon.Outcome, "years") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")) * 365,
    str_detect(Age.upon.Outcome, "months") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")) * 30,
    str_detect(Age.upon.Outcome, "weeks") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")) * 7,
    str_detect(Age.upon.Outcome, "days") ~ as.numeric(str_extract(Age.upon.Outcome, "\\d+")),
    TRUE ~ NA_real_
  ))
head(dogs.success.outcomes.age)
```

```r
dogs.success.outcomes.sex = dogs.success.outcomes.age %>%
  mutate(Sex.Outcome = ifelse(Sex.upon.Outcome %in% c("Neutered Male", "Intact Male"), 2, 3))
head(dogs.success.outcomes.sex)
```

```r
cats.success.outcomes.sex = cats.success.outcomes.age %>%
  mutate(Sex.Outcome = ifelse(Sex.upon.Outcome %in% c("Neutered Male", "Intact Male"), 2, 3))
head(cats.success.outcomes.sex)
```

```r
cats.before <- cats.success.outcomes.sex %>%
  mutate(Breed.Group = ifelse(str_detect(Breed, "/"), "Crossbreed", "Purebred")) %>%
  mutate(Breed.Group = ifelse(str_detect(Breed, "Mix"), "Crossbreed", Breed.Group))
head(cats.before)
```

```r
cats.middle <- cats.before %>%
    mutate(Breed1 = sapply(str_split(Breed, "/"), "[", 1)) %>%
    mutate(Breed2 = sapply(str_split(Breed, "/"), "[", 2))
head(cats.middle)
```

```r
dogs.before <- dogs.success.outcomes.sex %>%
  mutate(Breed.Group = ifelse(str_detect(Breed, "/"), "Crossbreed", "Purebred")) %>%
  mutate(Breed.Group = ifelse(str_detect(Breed, "Mix"), "Crossbreed", Breed.Group))
head(dogs.before)
```

```r
dogs.middle <- dogs.before %>%
    mutate(Breed1 = sapply(str_split(Breed, "/"), "[", 1)) %>%
    mutate(Breed2 = sapply(str_split(Breed, "/"), "[", 2))
head(dogs.middle)
```

```r
dogs.end = dogs.middle %>%
  mutate(Breed.Stat = ifelse(Breed.Group %in% c("Purebred"), 4, 5))
head(dogs.end)
```

```r
cats.end = cats.middle %>%
  mutate(Breed.Stat = ifelse(Breed.Group %in% c("Purebred"), 4, 5))
head(cats.end)
```

```r
library(ggplot2)
ggplot(cats.end, aes(x = Breed1, fill = factor(Successful.Outcome))) +
  geom_histogram(stat = "count", position = "dodge") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
  labs(title = "Histogram of Breed 1 with Successful Outcome",
       x = "Breed 1",
       y = "Count",
       fill = "Successful Outcome")
```

```r
library(ggplot2)
ggplot(cats.end, aes(x = Breed.Group, fill = factor(Successful.Outcome))) +
  geom_histogram(stat = "count", position = "dodge") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
  labs(title = "Histogram of Breed1 with Successful Outcome",
       x = "Breed Group",
       y = "Count",
       fill = "Successful Outcome")
```

```r
c.breed.results <- cats.end %>%
  group_by(Breed.Group) %>%
  summarise(
    p_value = {
      contingency_table <- table(Successful.Outcome)

      test_result <- chisq.test(contingency_table)
      test_result$p.value
    }
  )
print(c.breed.results)
```

```r
c.breed.results %>%
  mutate(
    significant = ifelse(p_value < 0.05, "Yes", "No")
  )
```

```r
cat_contingency_table <- table(cats.end$Breed.Group, cats.end$Successful.Outcome)
print(cat_contingency_table)
```
